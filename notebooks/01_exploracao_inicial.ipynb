## 6. Valores ausentes e colunas constantes

Na análise inicial, foram identificadas colunas com variância zero (`EmployeeCount`, `StandardHours`, `Over18`) — candidatas à remoção em etapas posteriores.

```python
ausentes = df.isnull().sum()
print("Valores ausentes por coluna:")
print(ausentes[ausentes > 0] if ausentes.sum() > 0 else "Nenhum valor ausente.")
print(f"\nTotal de valores ausentes: {ausentes.sum()}")
```

```python
nunicos = df.nunique().sort_values()
print("Número de valores únicos por coluna (menores primeiro):\n")
print(nunicos)

constantes = nunicos[nunicos == 1].index.tolist()
print("\nColunas constantes (variância zero):", constantes if constantes else "nenhuma")
```

## 7. Distribuição da variável-alvo (Attrition)
